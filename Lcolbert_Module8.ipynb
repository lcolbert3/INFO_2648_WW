1. Create a multi-line variable with the following dialogue from the book "Alice in Wonderland".

Have some wine,” the March Hare said in an encouraging tone.
Alice looked all round the table, but there was nothing on it but tea. “I don’t see any wine,” she remarked.
“There isn’t any,” said the March Hare.
“Then it wasn’t very civil of you to offer it,” said Alice angrily.
“It wasn’t very civil of you to sit down without being invited,” said the March Hare.
“I didn’t know it was your table,” said Alice; “it’s laid for a great many more than three.”

2.  Split and display only the lines for Alice. 
3. Split and display only the lines for the Hare.
4. Display only the second line spoken by Alice where she speaks angrily at the Hare.
5. Use regular expressions (regex) to substitute "angrily" for "sharply". NOTE: There maybe a few ways to do this. I recommend using re.sub module which finds a pattern (in this case the pattern is the word "angrily") and replaces it with something else.

```python
dialogue = '''
Have some wine,” the March Hare said in an encouraging tone.
Alice looked all round the table, but there was nothing on it but tea. “I don’t see any wine,” she remarked.
“There isn’t any,” said the March Hare.
“Then it wasn’t very civil of you to offer it,” said Alice angrily.
“It wasn’t very civil of you to sit down without being invited,” said the March Hare.
“I didn’t know it was your table,” said Alice; “it’s laid for a great many more than three.”
'''
```

```python
print(dialogue)
```

```python
d1 = dialogue.split('.')
```

```python
alice_lines = [s for s in d1 if "Alice" in s]
```

```python
for line in alice_lines:
    print(line)
```

```python
hare_lines = [h for h in d1 if 'Hare' in h]
```

```python
print(hare_lines)
```

Display only the second line spoken by Alice where she speaks angrily at the Hare.

```python
alice_angrily = alice_lines[1]
```

```python
print(alice_angrily)
```

Use regular expressions (regex) to substitute "angrily" for "sharply". NOTE: There maybe a few ways to do this. I recommend using re.sub module which finds a pattern (in this case the pattern is the word "angrily") and replaces it with something else.

```python
import re
```

```python
pattern = 'angrily'
```

```python
replacement = 'sharply'
```

```python
alice_sharp = re.sub(pattern, replacement, alice_angrily)
```

```python
print(alice_sharp)
```
