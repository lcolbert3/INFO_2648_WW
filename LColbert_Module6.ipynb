Write a program using the attached datasets above by doing the following:

1. Display a sample of the attached data's first 5 line.
2. Melt/Pivot columns "Selling Price", "Original Price", "Operating System" and "Rating", then display the top 5 line and the bottom 5 lines. 
3. Display a filtered sample of the melted data showing only the TOSHIBA brand.

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("TV_Final.csv")
```

```python
df.head()
```

```python
MELT_COLS = ["Selling Price", "Original Price", "Operating System", "Rating"]
```

```python
def get_id_vars(df, melt_cols):
    """
    Return the columns in the df that are NOT melts
    these are used as the Identifier columns
    """

    return [col for col in df.columns if col not in melt_cols]
```

```python
def melt_price_os_rating(df, melt_cols=None):

    if melt_cols is None:
        melt_cols = MELT_COLS

    id_vars =  get_id_vars(df, melt_cols)

    melted_df = pd.melt(
    df, 
    id_vars= id_vars,
    value_vars=melt_cols,
    var_name="Attribute",
    value_name="Value"
    )
    return melted_df
```

```python
df_melted = melt_price_os_rating(df)
```
