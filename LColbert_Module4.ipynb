```python
#import statement for the data

import pandas as pd
```

```python
#reading in the meals
meal = pd.read_csv("Datasets/datasets/data/meal_info.csv")
```

```python
#reading in the test data
test = pd.read_csv("Datasets/datasets/data/test.csv")
        
```

```python
#importing the fullfillment data
fulfilment = pd.read_csv("Datasets/datasets/data/fulfilment_center_info.csv")
```

### Creating the many to many relationship using the two different datasets and merging on the two columns centerid and mealid

```python
many_merged_meals = meal_cost.merge(center_test, on = ["center_id", "meal_id"])
```
