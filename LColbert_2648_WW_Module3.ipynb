```python
import seaborn as sns
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np
```

```python
print(sns.get_dataset_names())

```

```python
diamonds = sns.load_dataset('diamonds')
```

```python
diamonds.head(3)
```

# First Scatterplot

```python

plt.figure(figsize=(8,6))
sns.scatterplot(data = diamonds, x = 'price', y = 'carat', hue= 'color')
```

# Second Scatterplot

```python
fg = sns.FacetGrid(
    data = diamonds,
    col = "cut",
    row = "color",
    height = 3,
    aspect=1.2, 
    hue= 'clarity'
)
fg.map_dataframe(
     sns.scatterplot,
    x = "price",
    y = "carat"
    
)

fg.add_legend()
plt.show()
```
